```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("../data/raw/football_matches_dataset.csv")

print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

# First 5 rows
display(df.head())

# Data types
print("\nData Types:")
print(df.dtypes)

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

# Convert date
df["Match_Date"] = pd.to_datetime(
    df["Match_Date"],
    errors="coerce"
)

# Check invalid dates
print("Invalid Dates:", df["Match_Date"].isnull().sum())

# Numerical columns
numeric_columns = df.select_dtypes(
    include=np.number
).columns

print("\nNumerical Columns:")
print(numeric_columns.tolist())

# Check negative values
for col in numeric_columns:
    negative_count = (df[col] < 0).sum()
    if negative_count > 0:
        print(col, "negative values:", negative_count)

# Statistical summary
display(df.describe())



print("Cleaned dataset saved successfully!")
```

```python

# Total Goals
df["Total_Goals"] = (
    df["Home_Goals"] +
    df["Away_Goals"]
)

# Goal Difference
df["Goal_Difference"] = (
    df["Home_Goals"] -
    df["Away_Goals"]
)

# Absolute Goal Difference
df["Absolute_Goal_Difference"] = (
    df["Goal_Difference"].abs()
)

# Total Shots on Target
df["Total_Shots_On_Target"] = (
    df["Shots_On_Target_Home"] +
    df["Shots_On_Target_Away"]
)

df["Possession_Difference"] = (
    df["Possession_Home_Percent"] -
    df["Possession_Away_Percent"]
)


df["Absolute_Possession_Difference"] = (
    df["Possession_Difference"].abs()
)


df["Year"] = df["Match_Date"].dt.year


df["Month"] = df["Match_Date"].dt.month


df["Day_Name"] = df["Match_Date"].dt.day_name()

# Display first 5 rows
display(df.head())

df.to_csv(
    "../data/processed/football_cleaned.csv",
    index=False
)
print("Cleaned dataset saved successfully!")
```
